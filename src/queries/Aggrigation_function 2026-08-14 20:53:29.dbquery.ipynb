```sql
/* Aggrigation Function
1. MIN
2. MAX
3. AVG
4. SUM
5. COUNT
6. HAVING
7. GROUP BY
8. ORDER BY
9. LIMIT
*/

SELECT * 
FROM rivadata_platform.landing.students
;

SELECT MIN(student_id) 
FROM rivadata_platform.landing.students
;

SELECT MAX(student_id)
FROM rivadata_platform.landing.students
;

SELECT avg(student_id) 
FROM rivadata_platform.landing.students
;

SELECT SUM(student_id) 
FROM rivadata_platform.landing.students
;

SELECT count (*) 
FROM rivadata_platform.landing.students
;

SELECT * 
FROM rivadata_platform.landing.students
Order BY student_id DESC
;

SELECT * 
FROM rivadata_platform.landing.students
Order BY email 
;

SELECT country, count(*) 
FROM rivadata_platform.landing.students
GROUP BY country
;

SELECT
    country,
    COUNT(*) AS student_count
FROM rivadata_platform.landing.students
GROUP BY country
HAVING country = 'USA'
;
```
