```sql

-- Create table batch
CREATE OR REPLACE TABLE rivadata_platform.dataproduct.dim_batch(
  batch_key BIGINT NOT NULL,
  batch_id	STRING NOT NULL,
  batch_name	STRING NOT NULL,
  start_date	DATE NOT NULL,
  end_date	DATE,
  status	STRING NOT NULL,
  created_at	TIMESTAMP NOT NULL,

  CONSTRAINT pk_dim_batch
    PRIMARY KEY (batch_key)
);

-- Insert into batch table
INSERT INTO rivadata_platform.dataproduct.dim_batch (
  batch_key,
  batch_id,
  batch_name,
  start_date,
  end_date,
  status,
  created_at
)

-- Query on table as per requirement 
SELECT
    ROW_NUMBER() OVER(
      ORDER BY CAST (batch_id AS BIGINT)
    ) AS batch_key,
    batch_id,
    batch_name,
    start_date,
    end_date,
    status,
    CAST (created_at AS TIMESTAMP)
FROM rivadata_platform.landing.batches;


 








```
